```python
# =====================================================
# IA DE YOUTUBERS VIRTUALES
# CBTIS 128 - Programacion e IA
# Uso de:
# - Lista de diccionarios
# - Diccionarios Key : Value
# - if, elif, else
# - while True
# - ciclo for
# - funciones
# =====================================================

print("======================================================")
print("          IA DE YOUTUBERS VIRTUALES")
print("           CBTIS 128 - Programacion")
print("======================================================")

# Lista vacia
youtubers = []


# =====================================================
# FUNCION PARA AGREGAR YOUTUBER
# =====================================================
def agregar_youtuber():
    print("\n===== REGISTRO DE YOUTUBER =====")

    nombre = input("Nombre del youtuber: ")
    categoria = input("Categoria de videos: ")
    suscriptores = input("Cantidad de suscriptores: ")
    placas = input("Placas obtenidas: ")
    software = input("Software de edicion: ")
    camara = input("Camara utilizada: ")
    patrocinador = input("Patrocinadores: ")

    # Diccionario
    youtuber = {
        "Nombre": nombre,
        "Categoria": categoria,
        "Suscriptores": suscriptores,
        "Placas": placas,
        "Software": software,
        "Camara": camara,
        "Patrocinador": patrocinador
    }

    # Agregar a la lista
    youtubers.append(youtuber)

    print("\n✅ Youtuber agregado correctamente.")


# =====================================================
# FUNCION MOSTRAR YOUTUBERS
# =====================================================
def mostrar_youtubers():

    print("\n================ LISTA DE YOUTUBERS ================")

    if len(youtubers) == 0:
        print("No hay youtubers registrados.")
    else:

        print("-" * 130)

        print(
            f"{'NOMBRE':15}"
            f"{'CATEGORIA':18}"
            f"{'SUSCRIPTORES':15}"
            f"{'PLACAS':12}"
            f"{'SOFTWARE':18}"
            f"{'CAMARA':20}"
            f"{'PATROCINADOR':20}"
        )

        print("-" * 130)

        # Ciclo for
        for y in youtubers:

            print(
                f"{y['Nombre']:15}"
                f"{y['Categoria']:18}"
                f"{y['Suscriptores']:15}"
                f"{y['Placas']:12}"
                f"{y['Software']:18}"
                f"{y['Camara']:20}"
                f"{y['Patrocinador']:20}"
            )

        print("-" * 130)


# =====================================================
# FUNCION BUSCAR YOUTUBER
# =====================================================
def buscar_youtuber():

    nombre_buscar = input("\nEscribe el nombre del youtuber: ")

    encontrado = False

    for y in youtubers:

        if y["Nombre"].lower() == nombre_buscar.lower():

            print("\n=========== YOUTUBER ENCONTRADO ===========")

            print("Nombre:", y["Nombre"])
            print("Categoria:", y["Categoria"])
            print("Suscriptores:", y["Suscriptores"])
            print("Placas:", y["Placas"])
            print("Software:", y["Software"])
            print("Camara:", y["Camara"])
            print("Patrocinador:", y["Patrocinador"])

            encontrado = True

    if encontrado == False:
        print("❌ Youtuber no encontrado.")


# =====================================================
# FUNCION ANALISIS IA
# =====================================================
def analisis_ia():

    nombre = input("\nNombre del youtuber para analizar: ")

    encontrado = False

    for y in youtubers:

        if y["Nombre"].lower() == nombre.lower():

            print("\n============= ANALISIS IA =============")

            # Condicionales
            if y["Categoria"].lower() == "gaming":
                print("🎮 IA: Canal con gran potencial en Twitch y YouTube Shorts.")

            elif y["Categoria"].lower() == "tecnologia":
                print("💻 IA: Excelente categoria para conseguir patrocinadores.")

            elif y["Categoria"].lower() == "humor":
                print("😂 IA: El contenido de humor puede hacerse viral facilmente.")

            elif y["Categoria"].lower() == "musica":
                print("🎵 IA: Buen nicho para colaboraciones musicales.")

            else:
                print("📹 IA: Categoria creativa con oportunidades de crecimiento.")

            # Segunda evaluacion
            if y["Placas"].lower() == "diamante":
                print("🏆 Nivel elite de creador de contenido.")

            elif y["Placas"].lower() == "oro":
                print("🥇 Canal muy popular en la plataforma.")

            elif y["Placas"].lower() == "plata":
                print("🥈 Canal en crecimiento constante.")

            else:
                print("⭐ Sigue creando contenido para obtener placas.")

            encontrado = True

    if encontrado == False:
        print("❌ No existe ese youtuber.")


# =====================================================
# MENU PRINCIPAL
# =====================================================
while True:

    print("\n================ MENU PRINCIPAL ================")
    print("1. Registrar youtuber")
    print("2. Mostrar lista de youtubers")
    print("3. Buscar youtuber")
    print("4. Analisis IA del canal")
    print("5. Salir")

    opcion = input("Selecciona una opcion: ")

    # if elif else
    if opcion == "1":
        agregar_youtuber()

    elif opcion == "2":
        mostrar_youtubers()

    elif opcion == "3":
        buscar_youtuber()

    elif opcion == "4":
        analisis_ia()

    elif opcion == "5":
        print("\n👋 Saliendo del sistema IA de Youtubers...")
        break

    else:
        print("❌ Opcion invalida.")
```
